```python
import os
import zipfile
from kaggle.api.kaggle_api_extended import KaggleApi

# 1. Initialize and authenticate the API
api = KaggleApi()
api.authenticate()

# 2. Force Kaggle to download the dataset as a ZIP archive
# Notice we do NOT use the "-f sales_profit.csv" part here
dataset_slug = 'vikas605/sales-profit'
print("Downloading dataset archive...")
api.dataset_download_files(dataset_slug, path='.', unzip=False)

# 3. Kaggle names the downloaded zip after the end of the slug
zip_filename = 'sales-profit.zip'

# 4. Ankit Bansal's 2-line extraction code!
print(f"Extracting {zip_filename}...")
with zipfile.ZipFile(zip_filename, 'r') as zip_ref:
    zip_ref.extractall('.')

print("✨ Success! The zip file was downloaded and extracted. 'sales_profit.csv' is ready!")

```
